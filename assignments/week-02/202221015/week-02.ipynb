### 1-1. 텐서와 자동미분

2주차 온라인에서 본 **자동미분**의 동작을 숫자로 확인한다. `requires_grad=True` 텐서에 연산을 걸면 `backward()` 호출 시 그래디언트가 채워진다.


```python
import torch

# x 에 대해 loss 가 미분 가능한지 확인한다
x = torch.tensor(3.0, requires_grad=True)
w = torch.tensor(2.0, requires_grad=True)
b = torch.tensor(1.0, requires_grad=True)

loss = (w * x + b - 10) ** 2   # w*x + b 가 10 에 가까워지도록 하는 손실
loss.backward()

print(f"loss = {loss.item():.4f}")
print(f"d(loss)/dw = {w.grad.item():.4f}")
print(f"d(loss)/dx = {x.grad.item():.4f}")
print(f"d(loss)/db = {b.grad.item():.4f}")
```

### 1-2. 학습 루프 한 바퀴

이제 **순전파 - 손실 - 역전파 - 갱신** 순서를 한 바퀴 돌린다. 가중치가 한 번 갱신된 뒤 손실이 어떻게 변하는지 본다.


```python
import torch

# 데이터: x 가 주어졌을 때 3*x + 1 에 가까운 값을 내는 선형 모델을 맞춘다
x = torch.tensor([[1.0], [2.0], [3.0], [4.0]])
y = torch.tensor([[4.0], [7.0], [10.0], [13.0]])  # 정답: 3*x + 1

w = torch.tensor([[0.5]], requires_grad=True)
b = torch.tensor([[0.0]], requires_grad=True)

def predict(x):
    return x @ w + b

def mean_squared_error(pred, target):
    return ((pred - target) ** 2).mean()

lr = 0.01

for step in range(200):
    pred = predict(x)
    loss = mean_squared_error(pred, y)
    loss.backward()
    with torch.no_grad():
        w -= lr * w.grad
        b -= lr * b.grad
        w.grad.zero_()
        b.grad.zero_()
    if step % 40 == 0:
        print(f"step {step:3d}  loss = {loss.item():.6f}")

print(f"학습 후 w = {w.item():.3f}, b = {b.item():.3f}  (정답: w=3, b=1)")

with torch.no_grad():
    final_error = predict(x) - y
    result_mse = {'w': w.item(), 'b': b.item(), 'MSE': final_error.square().mean().item(), 'MAE': final_error.abs().mean().item()}
print('MSE 학습 후 공통 지표:', result_mse)
```

## 3. 실무: 방언·표준어 표기와 토크나이저 비용

[실무] 같은 뜻을 담는 데 드는 토큰 비용이 **표기에 따라** 달라지는지 실측한다. 사전학습 토크나이저는 표준어 말뭉치로 학습됐으므로, 방언 표기(제주어)는 같은 뜻의 표준어보다 **토큰 수가 더 많아질** 가능성이 크다.


### 3-1. 먼저 샘플로 감 잡기

데이터를 올리기 전에, 방언과 표준어가 토크나이저에서 어떻게 쪼개지는지 작은 예로 확인한다.


```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("monologg/koelectra-base-v3-discriminator", revision="68b30cd259f34a4b5aa8786392612ba2a2617fcc")

pairs = [
    ("혼저 옵서예.", "어서 오세요."),        # 방언 / 표준어
    ("마씀", "말씀"),
    ("고맙수다", "고맙습니다"),
]

for dialect, standard in pairs:
    d_tok = tokenizer.encode(dialect, add_special_tokens=False)
    s_tok = tokenizer.encode(standard, add_special_tokens=False)
    print(f"방언  [{dialect}] -> {len(d_tok)} 토큰 : {tokenizer.convert_ids_to_tokens(d_tok)}")
    print(f"표준  [{standard}] -> {len(s_tok)} 토큰 : {tokenizer.convert_ids_to_tokens(s_tok)}")
    print()
```

1. **손실 함수 변경 결과**

   같은 초기값, 학습률 0.01, 200회 갱신 조건에서 MSE 학습의 손실은 0회 60.375, 40회 0.002392, 160회 0.001152로 감소. MAE 학습은 0회 7.25, 40회 4.350002, 80회 1.449999, 120회 약 0.000005로 감소. MSE는 큰 오차에 큰 기울기를 주어 초기에 빠르게 감소하고, 작은 오차에서는 기울기도 줄어드는 특성. MAE는 오차의 크기보다 부호에 따라 기울기를 정하므로 초기에 거의 일정한 속도로 감소하는 양상.

   서로 다른 손실 값의 크기만으로 우열을 정할 수 없으므로, **마지막 갱신 후 동일한 두 지표**로 재계산한 결과:

   | 학습 목적 | 최종 w | 최종 b | 공통 MSE | 공통 MAE |
   |---|---:|---:|---:|---:|
   | MSE | 3.025053 | 0.926340 | 0.000906172 | 0.0250533 |
   | MAE | 3.000002 | 0.999999 | 3.50724e-11 | 5.36442e-06 |

   이 작은 선형 데이터와 지정한 200회 시점에서는 MAE 학습이 정답에 더 가까운 결과. 모든 데이터에서 MAE가 더 낫다는 근거는 아님. 고정 학습률의 MAE는 최솟값 가까이에서도 일정한 크기의 갱신이 일어나므로 진동 가능. 이상치, 학습률, 종료 시점에 따라 결과가 달라질 수 있음.

2. **역전파와 갱신의 역할**

   `loss.backward()`는 연쇄법칙으로 현재 손실을 각 학습 파라미터로 미분하여 `.grad`에 누적하는 단계. `w -= lr * w.grad`는 계산된 기울기의 반대 방향으로 파라미터 값을 실제 변경하는 단계. 다음 반복 전에 `zero_()`를 호출하는 이유는 이전 반복의 기울기가 의도치 않게 누적되는 것을 막기 위함. 첫 자동미분 예제의 손실 9, w/x/b의 기울기 -18/-12/-6도 실제 출력으로 확인.

3. **방언과 토큰 비용**

   `혼저 옵서예.` 6개 대 `어서 오세요.` 4개로 +2개, `마씀` 2개 대 `말씀` 1개로 +1개, `하르방` 2개 대 `할아버지` 1개로 +1개 확인. 반면 `고맙수다`와 `고맙습니다`는 각 3개, `쉰 살`은 양쪽 모두 2개. 방언이면 항상 늘어나는 것은 아님.

   토크나이저는 의미가 같은지를 먼저 판단하는 대신 학습된 어휘와 하위 단위 분절 규칙에 따라 표기를 분해함. 어휘에 잘 포함된 표기는 긴 단위로, 드문 표기는 더 작은 조각으로 나뉠 수 있어 같은 뜻에도 입력 토큰 수 차이가 생김. 제공된 소수 예시의 관찰이며 AI Hub 제주도 원본 말뭉치 전체의 통계로 일반화할 수 없음.
